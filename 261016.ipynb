```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
```

```python
# MNIST 데이터셋 불러오기
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# 이미지와 정답의 개수 확인
print("학습 이미지:", x_train.shape)
print("학습 정답:", y_train.shape)

print("테스트 이미지:", x_test.shape)
print("테스트 정답:", y_test.shape)
```

```python
# 0번째 이미지 선택
image = x_train[0]

# 정답 선택
label = y_train[0]

# 정답 확인
print("정답:", label)

# 이미지 출력
plt.imshow(image, cmap='gray')
plt.title(f"Number: {label}")
plt.axis("off")
plt.show()
```

```python
# 이미지 픽셀값 이해하기

print(image.shape)
print(image)
```

```python
# 0~225 범위를 0~1 범위로 변경
image = image.astype(np.float32) / 255.0

print("가장 작은 값:", image.min())
print("가장 큰 값:", image.max())
```

```python
# 패딩 구하기

# 패딩 함수 만들기
def add_padding(img):

    padded = np.pad(
        img,
        pad_width=1,
        mode="constant",
        constant_values=0
    )

    return padded


# 이미지 한 장에 패딩 적용
padded = add_padding(image)

print("원본:", image.shape)
print("패딩:", padded.shape)
```

```python
plt.imshow(padded, cmap="grey", vmin=0, vmax=1)
plt.title("Padding")
plt.axis("off")
plt.show()
```

```python
# 커널 구현하기

# 커널이란 ? 커널은 이미지에 계산을 적용할 때 사용하는 작은 숫자 배열

# 3x3 커널 만들기
kernel = np.ones((3, 3), dtype=np.float32) / 9

print(kernel)
```

```python
# 실제 적용 함수 만들기
def apply_kernel(img):

    # 이미지 높이와 너비 확인
    height, width = img.shape

    # 결과를 저장할 배열
    output = np.zeros(
        (height -2 , width -2),
        dtype=np.float32
    )

    # 이미지의 세로 방향 이동
    for row in range(height -2):

        # 이미지의 가로 방향 이동
        for col in range(width -2):

            # 3x3 영역 가져오기
            area = img[row:row+3, col:col+3]

            # 커널 곱셈 및 합계
            result = np.sum(area * kernel)

            # 결과 저장
            output[row, col] = result

    return output


# 패딩된 이미지에 커널 적용
filtered = apply_kernel(padded)
print("커널 적용 후:", filtered.shape)
```

```python
plt.imshow(filtered, cmap="gray", vmin=0, vmax=1)
plt.title("Kenrel")
plt.axis("off")
plt.show()
```

```python
# 최대 풀링 함수
def max_pooling(img):

    # 이미지 높이와 너비
    height, width = img.shape

    # 결과 배열 생성
    output = np.zeros(
        (height // 2, width // 2),
        dtype=np.float32
    )

    # 세로 방향
    for row in range(height // 2):

        # 가로 방향
        for col in range(width // 2):

            # 2×2 영역 가져오기
            area = img[
                row*2:row*2+2,
                col*2:col*2+2
            ]

            # 가장 큰 값 선택
            output[row, col] = np.max(area)

    return output


# 커널 적용 이미지에 풀링 적용
pooled = max_pooling(filtered)

print("풀링 적용 후:", pooled.shape)
```

```python
plt.imshow(pooled, cmap="gray", vmin=0, vmax=1)
plt.title("Max Pooling")
plt.axis("off")
plt.show()
```

```python
# 비교할 이미지
pictures = [
    image,
    padded,
    filtered,
    pooled
]

# 이미지 제목
titles = [
    "Original 28x28",
    "Padding 30x30",
    "Kernel 28x28",
    "Pooling 14x14"
]

# 네 개 이미지 공간 만들기
fig, axes = plt.subplots(1, 4, figsize=(12, 3))

for i in range(4):

    axes[i].imshow(
        pictures[i],
        cmap="gray",
        vmin=0,
        vmax=1
    )

    axes[i].set_title(titles[i])
    axes[i].axis("off")

plt.tight_layout()

# 결과 이미지 파일로 저장
plt.savefig("mnist_processing.png", dpi=150)

plt.show()
```

```python
# MNIST 이미지 100장 선택
images = x_train[:100].astype(np.float32) / 255.0

# 각 이미지의 정답
labels = y_train[:100]

# 결과 저장할 빈 리스트
padding_list = []
kernel_list = []
pooling_list = []

# 이미지 100장을 하나씩 처리
for img in images:

    # 1. 패딩
    pad_img = add_padding(img)

    # 2. 커널 적용
    kernel_img = apply_kernel(pad_img)

    # 3. 풀링
    pool_img = max_pooling(kernel_img)

    # 처리된 이미지 저장
    padding_list.append(pad_img)
    kernel_list.append(kernel_img)
    pooling_list.append(pool_img)


# 리스트를 NumPy 배열로 변경
padding_images = np.array(padding_list)
kernel_images = np.array(kernel_list)
pooling_images = np.array(pooling_list)

print("원본:", images.shape)
print("패딩:", padding_images.shape)
print("커널:", kernel_images.shape)
print("풀링:", pooling_images.shape)
```

```python
# 팀원별로 이름 설정
student_name = "member_A"

# 데이터셋 저장
np.savez_compressed(
    f"{student_name}_mnist_processed.npz",

    images=images,
    labels=labels,
    padding=padding_images,
    kernel=kernel_images,
    pooling=pooling_images
)

print("데이터셋 저장 완료!")
```

```python
# 저장한 데이터셋 불러오기
data = np.load("member_A_mnist_processed.npz")

# 파일에 저장된 항목 확인
print(data.files)

# 풀링 이미지 크기 확인
print(data["pooling"].shape)

# 저장된 첫 번째 이미지 출력
plt.imshow(
    data["pooling"][0],
    cmap="gray",
    vmin=0,
    vmax=1
)

plt.title("Saved Pooling Image")
plt.axis("off")
plt.show()
```
